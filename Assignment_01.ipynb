```python
cf = int(input("enter current floor:"))
tf = int(input("enter target floor:"))

if tf == 13:
    print("Floor 13 is skipped. Redirecting to floor 14.")
    tf = 14

if cf < tf:
    r = tf - cf
    print(f"going up to {r} floors")
elif cf > tf:
    d = cf - tf
    print(f"going down to {d} floors")
else:
    print("You are on the same floor.")
```

```python
age = int(input("Enter age: "))
ticketp = 120

if age > 0:
    if age <= 12:
        weekend = input("Weekend? Yes/No: ")
        if weekend == "Yes":
            ticketp = (120 / 2) + 30
        else:
            ticketp = 120 / 2
        print(f"{ticketp} baht")
    elif age < 60:
        weekend = input("Weekend? Yes/No: ")
        if weekend == "Yes":
            ticketp = 120 + 30
        else:
            ticketp = 120
        print(f"{ticketp} baht")
    else:
        weekend = input("Weekend? Yes/No: ")
        if weekend == "Yes":
            ticketp = 80 + 30
        else:
            ticketp = 80
        print(f"{ticketp} baht")
else:
    print("Invalid age")
```

```python
b = int(input("Balance: "))
a = int(input("Amount: "))

if a > b:
    print("Insufficient balance")
elif a % 100 != 0:
    print("Amount must be a multiple of 100")
elif a >= 20000:
    print("Over daily limit")
else:
    nb = b - a
    print(f"Dispensing {a} baht.")
    print(f"New Balance: {nb} baht")
```

```python
units = float(input("Enter units: "))

if units <= 150:
    tb = units * 3
elif units <= 400:
    tb = 150 * 3 + (units - 150) * 4
else:
    tb = 150 * 3 + 250 * 4 + (units - 400) * 5

print(f"Total Bills: {tb} baht")
```
