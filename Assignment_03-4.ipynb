### The specification

Amounts are in **Thai baht (฿)**; meter readings are whole **units (kWh)**.

**1. `units_used(previous, current)`** → units consumed. Raise `ValueError` if either reading is negative, or if `current < previous`.

| 2 | next 250 (151–400) | 4.00 |
| 3 | 401 and above | 4.50 |

**7. `round_baht(amount)`** → `amount` rounded to 2 decimals.

**8. `total_due(units, rate_type, ft_rate, vat_rate=0.07)`** → `round_baht(subtotal + vat(subtotal))`.

**9. `format_baht(amount)`** → string like `฿1,759.08` (baht sign, thousands separator, 2 decimals).
