## Part 2 — Activity 2: Password Strength — Count the Digits  *(solve by yourself)*

**task 1:** many sign-up forms check that a password contains enough numbers. Ask for a password and count how many **digits** (0-9) it contains.

**Use:** a `for` loop, a **counter**, and the `in` operator. Do **not** use a function this week.

*Sample run:*
```
Enter a password: p@ss2024
Number of digits: 4
```

*Hints:*
- Start with `digits = '0123456789'` and `count = 0`.
- Loop over each character: `for ch in password:`
- If `ch in digits`, add 1 to `count`.

*If you finish quickly:* also print `Looks OK` when the password has **at least 2 digits**, otherwise `Too few numbers`.

````python
count = 0
password = input("Enter password:")
for ch in password:
    if ch.isdigit():
        count += 1

if count >= 2:
    print("LOOKS OK")

else:
    print("T0O FEW NUMBERS")

   


````

## Part 3 — Optional stretch: Login with Limited Attempts

**task 1:** when you type a PIN or password, the system gives you a few tries and then locks you out. Simulate it.

- The correct PIN is `'2468'` (just for this exercise).
- Let the user try up to **3 times**.
- After each wrong try, print how many attempts are left.
- Print `Access granted.` on success, or `Account locked.` after 3 wrong tries.

*Sample run:*
```
Enter your PIN: 0000
Wrong PIN. Attempts left: 2
Enter your PIN: 1111
Wrong PIN. Attempts left: 1
Enter your PIN: 2468
Access granted.
```

*Hint:* a `while` loop that continues **while the PIN is wrong AND attempts remain**. Read once before the loop, then again at the end of each pass. Change `attempts` each pass, or it will loop forever.

````python
count = 3
attempt = count
cpin = 3333
while count > 0:
      pin = int(input('Enter pin:'))
      if pin == cpin:
         print("Access granted")
         break
      else:
           attempt -= 1
           print(f"Wrong Pin. Attempts left: {attempt}")
           if attempt == 0:
              print("Account Locked")
              break

````

## Part 4 — Average Download Speed

**task:** a network tool samples the download speed again and again until you tell it to stop, then reports the average. You do **not** know in advance how many samples there will be.

- Read speeds (Mbps) one at a time. Stop when the user enters `0`.
- Print how many readings were taken and the **average**, rounded to 2 decimals.
- If the first entry is `0`, print `No readings.`

*Sample run:*
```
Download speed Mbps (0 to stop): 45
Download speed Mbps (0 to stop): 55
Download speed Mbps (0 to stop): 50
Download speed Mbps (0 to stop): 0
Readings: 3
Average speed: 50.0 Mbps
```

This is a **sentinel loop** — the same pattern you will need in this week's assignment. Keep a running total and a counter, and watch the empty case.

````python
reading = 0
total = 0
avg = 0
while True:
      speed = int(input("Enter download speed:"))
      if speed >= 1:
         total += speed
         reading += 1
         avg = total / reading
         print(f"Download speed Mbps (o to stop): {speed}")
      else:
           print("No readings")
           break
print(f"Reading: {reading}")
print(f"Average speed: {avg:.2f}")
            
````
