## Function 2 — `calculate_fare(seat, age, bags)`

- `seat` — a **valid, cleaned** seat code, e.g. `'3F'`
- `age` — passenger age in years (int)
- `bags` — number of checked bags (int, 0 or more)

1. Base fare is **2500 THB**.
2. **Business** class (rows 1–5): multiply the base fare by **2.5**.
3. **Window** seat (A or F): add **200 THB**.
4. **Age discount** on the seat price from steps 1–3:

5. **Bags:** the **first bag is free**; every extra bag costs **500 THB**. Bag fees are **not** discounted.

````python


def is_valid_seat(seat, booked): #set the function
    seat = seat.strip().upper() #clean spaces and make the value upper case 
    row = seat[:-1] #slices all except the last one
    if len(seat) < 2 or len(seat) > 3: #use len to count seat length to compare to integer
        return 'invalid format' #on condt return this
    if not seat[-1].isalpha(): #grab the last individual character check if it is not a letter
        return 'invalid format' #return this on condt
    if not row.isdigit(): #check if the numbers before the letter are not digit
        return 'invalid format' #return
    if int(row) < 1 or int(row) > 30: #use int to convert row into interger anc set compare 
       return 'row out of range' #return
    if seat[-1] not in 'ABCDEF': #check if the last character is not in abcdef
         return 'invalid letter' #return
    if seat in booked: #if what user enter is already in booked list or not
        return'already booked' #return
    
    return 'OK' #else return(after passing above condt)


def calculate_fare(seat, age, bags): #set condt
    base_fare = 2500 #set initial fare
    row = int(seat[:-1]) #set row to one or two numbers before the letter
    fare = base_fare #set fare == base_fare

    if 1 <= row <= 5: #check row is in business
        fare = fare * 2.5 #if true multiplied 2500 by 2.5

    if seat[-1] in 'AF': #check if is window seat
        fare = fare + 200 #if true add 200 to the fare

    if age < 2: #check infant
        fare = fare * 0.10 #if true calculate 10% of fare 
    elif age <= 11: #check kid or not
        fare = fare * 0.75 #if true calculate 75% of fare
    if bags > 1: #check if there is more than 1 bag
            fare = fare + (bags - 1) * 500 #if true find extra bags and multiplied it by 500 then add to the fare

    #the fare will accumulate based on condt
    #initial price is 2500 

    return round(fare, 2)#return accumulated fare

def create_ticket(name, flight, seat, fare): #set condt

    words = name.split()
    word = ' '.join(words)

    line1 = f"============ E-TICKET ============" #line1 eqauls to this lomg format
    line2 = f"Passenger  :  {word.title()}" #line2 = name that is stripped of spaces and titled
    line3 = f"Flight  :  {flight}" #line3 = flight number
    if  1 <= int(seat[:-1]) <=5: #check if seat is in business or not
        line4 = f"Seat   :   {seat} (Business)" #trigger uppon business
    else:
        line4 =f"Seat   :   {seat} (Economy)"#trigger uppon else 
    line5 = f"Fare     : {fare:,.2f} THB" #formated fare seperated by ',' and rouned up to 2 decimal
    line6 = f"Booking  :  {words[-1][:3].strip().upper()}-{flight.upper()}-{seat.upper()} -> CNX" #3 chr of names stripped of spaces make it uppercase - flght is uppercase - seat is uppercase
    combined_line = line1 + "\n" + line2 + "\n" + line3 + "\n" + line4 + "\n" + line5 + "\n" + line6 #combines the lines above
    return combined_line #return the result and stored


# ----------------------------- MAIN PROGRAM -----------------------------
FLIGHT = 'TG105' #set flight
booked = ['1A', '12C', '20F'] #set booked list to compare in is_valid_seat
tickets = 0 #set tickets
revenue = 0 #set revenue
while True: #if true run inner 
    name = input("Enter passenger name or done: ") #while true keep asking name
    if name.strip().lower() == 'done': #clean spaces, make the input lower case and compare to done18
       break #if it is done then break the loop
    while True: #inner while, only run while outer while activate
        seat = input("Seat: ") #ask input
        result = is_valid_seat(seat, booked) #result stored whatever the first function return( if first function pass all the condt = OK, else = return invalid, out of range......)
        if result == 'OK': #result = OK #check if result is OK
            break #if true break the loop
        print("x", result) #else print x

    age = int(input("Age: ")) #ask age while true
    bags = int(input("Bags: ")) #ask bag numbers while true

    seat = seat.strip().upper()          # cleaned seat
    fare = calculate_fare(seat, age, bags) #call function while true

    booked.append(seat) #append the booked list on successful
    tickets = tickets + 1 #increase the count
    revenue = revenue + fare #add to revenue afteer 2nd function is called and calculated 


    receipt = create_ticket(name, FLIGHT, seat, fare)   #call 3rd function to print
    print(receipt)


print(f"\nTotal tickets sold: {tickets}")       #print total tickets
print(f"Total revenue: {revenue:,.2f} THB") #print revenue, formated ',' and rounded up to 2 decimal
BOOKED = ', '.join(booked) #add items from lists to this using ','
print(BOOKED)  #print the items of booked 
# Write your main program here
````
